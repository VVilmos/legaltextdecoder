```python
import pandas
import plotly.express as px
import matplotlib.pyplot as plt
```

```python
df = pandas.read_csv("./data/raw_data.csv")
```

## Dropping duplicates

```python
print(f"Number of duplicated rows: {df.duplicated().sum()}")
print(f"Number of missing values:\n{df.isnull().sum()}")
```

# Analysing paragraph features

## Word count and paragraph length distribution

### Summary

```python
# Calculate paragraph lengths
df['paragraph_length'] = df['paragraph'].str.len()
df['word_count'] = df['paragraph'].str.split().str.len()

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Character length distribution
axes[0].hist(df['paragraph_length'], bins=50, color='skyblue', edgecolor='black', alpha=0.7)
axes[0].set_title('Distribution of Paragraph Length (Characters)', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Character Count', fontsize=11)
axes[0].set_ylabel('Frequency', fontsize=11)
axes[0].axvline(df['paragraph_length'].mean(), color='red', linestyle='--', linewidth=2, label=f'Mean: {df["paragraph_length"].mean():.0f}')
axes[0].axvline(df['paragraph_length'].median(), color='green', linestyle='--', linewidth=2, label=f'Median: {df["paragraph_length"].median():.0f}')
axes[0].legend()

# Word count distribution
axes[1].hist(df['word_count'], bins=50, color='lightcoral', edgecolor='black', alpha=0.7)
axes[1].set_title('Distribution of Word Count', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Word Count', fontsize=11)
axes[1].set_ylabel('Frequency', fontsize=11)
axes[1].axvline(df['word_count'].mean(), color='red', linestyle='--', linewidth=2, label=f'Mean: {df["word_count"].mean():.0f}')
axes[1].axvline(df['word_count'].median(), color='green', linestyle='--', linewidth=2, label=f'Median: {df["word_count"].median():.0f}')
axes[1].legend()

plt.tight_layout()
plt.show()
```

### Grouped by classes

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Box plot for character length by label
df.boxplot(column='paragraph_length', by='label', ax=axes[0])
axes[0].set_title('Paragraph Length (Characters) by Label', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Label', fontsize=11)
axes[0].set_ylabel('Character Count', fontsize=11)
plt.sca(axes[0])
plt.xticks(rotation=0)

# Box plot for word count by label
df.boxplot(column='word_count', by='label', ax=axes[1])
axes[1].set_title('Word Count by Label', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Label', fontsize=11)
axes[1].set_ylabel('Word Count', fontsize=11)
plt.sca(axes[1])
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```
